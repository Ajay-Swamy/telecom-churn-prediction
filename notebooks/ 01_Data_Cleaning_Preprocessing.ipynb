## 5. Missing Value Analysis

```python
(df.isnull().sum().sort_values(ascending=False)/len(df))*100
```

```python
df[df['TotalCharges'].isnull()]
# here we can observe that when tenure is 0 the total charges is null
# which means the customers with 0 tenure are not having total charges as they are New customers
```

```python
df[df["tenure"] == 0]
```

```python
df[df["tenure"] == 0].equals(df[df['TotalCharges'].isnull()])
```

- Missing Value Analysis
- Checked duplicate records — No duplicate rows found.
- Identified 11 missing values in TotalCharges.
- Investigation showed all missing values belonged to customers with tenure = 0.
- Based on business logic, these customers had not completed their first billing cycle, so their TotalCharges will be replaced with 0 during preprocessing.

```python
# Replace missing TotalCharges with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```
