# Telecom Customer Churn Analytics & Prediction

## Notebook 3: Feature Engineering & Data Preprocessing

### Objective

The objective of this notebook is to prepare the cleaned telecom dataset for machine learning.

The preprocessing steps include:

- Understanding feature engineering
- Dropping unnecessary columns
- Encoding categorical variables
- Feature scaling
- Separating features and target
- Train-Test Split
- Preparing the final dataset for machine learning algorithms

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('../Data/processed/telecom_churn_clean.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.drop('customerID',axis=1,inplace=True)
```

**The customerID column was removed because
it is a unique identifier and does not contribute meaningful information for predicting customer churn.
Including identifier columns can introduce noise without improving model performance.**

```python
for col in df.select_dtypes(include='object').columns:
    print(f'{col} --> {df[col].unique()}')
    print('-'*50)
```

```python
binary_cols = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]
```

## Apply Label Encoding

```python
le = LabelEncoder()

for col in binary_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df[binary_cols].head()
```

```python
for col in binary_cols:
    print(col, ":", df[col].unique())
```

## Label Encoding

Binary categorical variables were converted into numerical values using Label Encoding.

The following features were encoded:

- gender
- Partner
- Dependents
- PhoneService
- PaperlessBilling
- Churn (Target Variable)

Label Encoding was selected because these variables contain only two categories, making binary numerical representation appropriate without introducing ambiguity.

## One Hot Encoding

```python
multi_cols = [
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaymentMethod"
]

df = pd.get_dummies(
    df,
    columns=multi_cols,
    drop_first=True,
    dtype=int
)
```

```python
df.head()
```

```python
df.info()
```

```python
df.select_dtypes(include="object").columns
```

## One-Hot Encoding

Multi-category categorical variables were transformed using One-Hot Encoding with `pd.get_dummies()`.

The following features were encoded:

- MultipleLines
- InternetService
- OnlineSecurity
- OnlineBackup
- DeviceProtection
- TechSupport
- StreamingTV
- StreamingMovies
- Contract
- PaymentMethod

The parameter `drop_first=True` was used to avoid the Dummy Variable Trap by removing one redundant category from each encoded feature. The resulting dataset contains only numerical variables, making it suitable for machine learning algorithms.

## Feature Scaling

```python
X = df.drop("Churn", axis=1)

y = df["Churn"]
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,
    test_size=0.20,
    random_state=42,
    stratify=y)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
numeric_cols = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
```

```python
scaler = StandardScaler()

X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])

X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])
```

```python
import joblib

joblib.dump(scaler, "../models/scaler.pkl")

print("Scaler saved successfully!")
```

```python
X_train.head()
```

## Data Preprocessing

The cleaned dataset was prepared for machine learning using the following steps:

- Removed the identifier column (`customerID`).
- Applied Label Encoding to binary categorical variables.
- Applied One-Hot Encoding to multi-category variables.
- Split the dataset into training (80%) and testing (20%) sets using stratified sampling.
- Standardized the numerical features (`tenure`, `MonthlyCharges`, and `TotalCharges`) using `StandardScaler` fitted only on the training data to prevent data leakage.

```python
processed_train = X_train.copy()
processed_train["Churn"] = y_train

processed_test = X_test.copy()
processed_test["Churn"] = y_test

processed_train.to_csv(
    "../Data/processed/train_data.csv",
    index=False
)

processed_test.to_csv(
    "../Data/processed/test_data.csv",
    index=False
)
```
